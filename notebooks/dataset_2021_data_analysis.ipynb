```python
df = pd.read_csv("~/apartment-price-prediction/data/raw/fetched/2021/input_data.csv", sep=";")
```

Построим и проанализируем график, где в качестве параметра будем смотреть заполнен столбец какими-то данными или NaN 

```python
plt.figure(figsize=(20,10))

data = df.isna().sum()
plt.bar(data.index, data.values)

plt.xlabel("Параметры")
plt.ylabel("Количество NaN")

plt.show()

print(df.isna().sum())
```
